```python
# Load env variables
from anthropic import Anthropic
from google.colab import userdata
```

```python
# 1. Retrieve the key from Colab Secrets
my_anthropic_key   = userdata.get('ANTHROPIC_API_KEY')
my_anthropic_model = userdata.get('CLAUDE_MODEL')
```

```python

# Create an API client
client = Anthropic(
    api_key=my_anthropic_key,
)

print("Anthropic client initialized successfully!")
```

```python
# define a system prompt
system_prompt = """
You are a creative screenwriter for feature films.
"""
```

```python
# Utility functions
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=1.0):
    params = {
        "model": my_anthropic_model,
        "max_tokens": 1000,
        "messages": messages
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
# Make a request

# Start with an empty message list
messages = []

# Add the initial user question
add_user_message(messages, "Generate a one sentence movie idea.")

# Get Claude's response
answer = chat(messages, system=system_prompt, temperature=0.0)

print("Answer:", answer)
```
